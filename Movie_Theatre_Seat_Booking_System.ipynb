- 0 means available
- 1 means booked

```python
seats = seats.astype(np.int64)
```

```python
for i, row in enumerate(seats):
  print(f"Row {i+1}: {row}")
```

```python
def display_seat():
  for i, row in enumerate(seats):
    print(f"Row {i+1}: {row}")
```

Seats Booking Algorithm:

```python
def seatBooking(row, col):
  if row < 1 or row > 5 or col < 1 or col > 8:
    print("Invalid seat number!")
    return
  if seats[row-1, col-1] == 1:
    print("Seat already booked.")
    return
  else:
    seats[row-1, col-1] = 1
    print("Seat booked successfully.")
```

```python
seatBooking(2,3)
```

```python
seatBooking(5,2)
```

```python
seatBooking(-1,2)
```
